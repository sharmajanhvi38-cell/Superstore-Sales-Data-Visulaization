```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_excel('/content/Superstore_Sales_Task3.xls')
df.head()
```

```python
print("Dataset Shape:")
print(df.shape)
print("\nColumnNames:")
for col in df.columns:
  print(col)
print("\nMissing Values:")
print(df.isnull().sum())
print("\nData Types:")
print(df.dtypes)
```

```python
# Convert Order Date into datetime format
df['Order Date'] = pd.to_datetime(df['Order Date'])

# Create monthly sales data
monthly_sales = (
    df.groupby(df['Order Date'].dt.to_period('M'))['Sales']
    .sum()
    .reset_index()
)

# Convert period to timestamp
monthly_sales['Order Date'] = monthly_sales['Order Date'].dt.to_timestamp()

# Plot
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly_sales,
    x='Order Date',
    y='Sales',
    marker='o'
)

plt.title('Monthly Sales Trend', fontsize=18, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
# Calculate total sales by category
category_sales = (
    df.groupby('Category')['Sales']
    .sum()
    .sort_values(ascending=False)
)

# Create bar chart
plt.figure(figsize=(9, 6))

sns.barplot(
    x=category_sales.values,
    y=category_sales.index
)

plt.title('Sales Performance by Category', fontsize=18, fontweight='bold')
plt.xlabel('Total Sales')
plt.ylabel('Category')

plt.tight_layout()
plt.show()
```

```python
# Calculate total sales by region
region_sales = (
    df.groupby('Region')['Sales']
    .sum()
    .sort_values(ascending=False)
)

# Create bar chart
plt.figure(figsize=(9, 6))

sns.barplot(
    x=region_sales.values,
    y=region_sales.index
)

plt.title('Sales Performance by Region', fontsize=18, fontweight='bold')
plt.xlabel('Total Sales')
plt.ylabel('Region')

plt.tight_layout()
plt.show()
```

```python
# Calculate total profit by category
category_profit = (
    df.groupby('Category')['Profit']
    .sum()
    .sort_values(ascending=False)
)

# Create bar chart
plt.figure(figsize=(9, 6))

sns.barplot(
    x=category_profit.values,
    y=category_profit.index
)

plt.title('Profit Performance by Category', fontsize=18, fontweight='bold')
plt.xlabel('Total Profit')
plt.ylabel('Category')

plt.tight_layout()
plt.show()
```

```python
# Calculate total sales by product
top_products = (
    df.groupby('Product Name')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .sort_values(ascending=True)
)

# Create horizontal bar chart
plt.figure(figsize=(11, 7))

sns.barplot(
    x=top_products.values,
    y=top_products.index
)

plt.title('Top 10 Products by Sales', fontsize=18, fontweight='bold')
plt.xlabel('Total Sales')
plt.ylabel('Product')

plt.tight_layout()
plt.show()
```

```python
# Sales vs Profit relationship
plt.figure(figsize=(10,7 ))

sns.scatterplot(
    data=df,
    x='Sales',
    y='Profit',
    hue='Category',
    alpha=0.6
)

plt.title('Sales vs Profit by Category', fontsize=18, fontweight='bold')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.axhline(0, linestyle='--', linewidth =1)

plt.tight_layout()
plt.show()
```

```python
# Calculate profit by sub-category
subcategory_profit = (
    df.groupby('Sub-Category')['Profit']
    .sum()
    .sort_values()
)

# Create horizontal bar chart
plt.figure(figsize=(11, 7))

sns.barplot(
    x=subcategory_profit.values,
    y=subcategory_profit.index
)

plt.title('Profitability by Sub-Category', fontsize=18, fontweight='bold')
plt.xlabel('Total Profit')
plt.ylabel('Sub-Category')
plt.axvline(0, linestyle='--', linewidth=1)

plt.tight_layout()
plt.show()
```

```python
# Select numerical columns
numeric_data = df[['Sales', 'Quantity', 'Discount', 'Profit']]

# Calculate correlation
correlation = numeric_data.corr()

# Create heatmap
plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Between Sales, Quantity, Discount and Profit',
          fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
df = pd.read_excel('/content/Superstore_Sales_Task3.xls')

# Generate key business metrics

total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_quantity = df['Quantity'].sum()

best_category = df.groupby('Category')['Sales'].sum().idxmax()
best_region = df.groupby('Region')['Sales'].sum().idxmax()
best_profit_category = df.groupby('Category')['Profit'].sum().idxmax()
best_product = df.groupby('Product Name')['Sales'].sum().idxmax()

print("===== BUSINESS INSIGHTS =====")

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Quantity Sold: {total_quantity:,}")

print(f"\nHighest Sales Category: {best_category}")
print(f"Highest Sales Region: {best_region}")
print(f"Most Profitable Category: {best_profit_category}")
print(f"Top Product by Sales: {best_product}")
```

Superstore Sales & Profit - Data Visulaization Project

Objective:
To transform raw sales data into meaningful visualizations, identify trends and patterns, analyze sales and profitability, and generate business insights for better decision-making.
Tools Used:
Python, Pandas, Matplotlib, Seaborn, Google Colab


Key Business Insights
Sales performance varies across different categories and regions, helping identify the strongest business areas.
High sales do not always result in high profit, making profitability analysis important for decision-making.
Top-performing products contribute significantly to overall sales.
Sub-category analysis helps identify profitable and low-performing product areas.
Sales, quantity, discount, and profit show different relationships that can help understand business performance.




Business Recommendations
Focus on high-performing categories and products.
Monitor low-profit and loss-making sub-categories.
Review discount strategies to protect profit margins.
Use regional and product-level performance to improve business decisions.
